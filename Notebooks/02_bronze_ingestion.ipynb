```python
ruta_principal="/Volumes/olist/default/raw/"
```

```python
Schema_bronze ='olist.bronze.'
#Customers
df_1 = spark.read.option("header", "true").option("inferSchema", "true").csv(ruta_principal+'olist_customers_dataset.csv')
df_1.write.format("delta").mode("overwrite").saveAsTable(Schema_bronze+'customers')

#Geolocation
df_2 = spark.read.option("header", "true").option("inferSchema", "true").csv(ruta_principal+'olist_geolocation_dataset.csv')
df_2.write.format("delta").mode("overwrite").saveAsTable(Schema_bronze+"geolocation")

#Order items
df_3 = spark.read.option("header", "true").option("inferSchema", "true").csv(ruta_principal+'olist_order_items_dataset.csv')
df_3.write.format("delta").mode("overwrite").saveAsTable(Schema_bronze+"order_items")

#Order payments
df_4 = spark.read.option("header", "true").option("inferSchema", "true").csv(ruta_principal+'olist_order_payments_dataset.csv')
df_4.write.format("delta").mode("overwrite").saveAsTable(Schema_bronze+"order_payments")

# Order reviews
df_5 = (
    spark.read
    .option("header", True)
    .option("inferSchema", True)
    .option("multiLine", True)
    .option("quote", '"')
    .option("escape", '"')
    .csv(ruta_principal + "olist_order_reviews_dataset.csv")
)

(
    df_5.write
    .format("delta")
    .mode("overwrite")
    .option("overwriteSchema", "true")
    .saveAsTable(Schema_bronze + "order_reviews")
)

#Orders
df_6 = spark.read.option("header", "true").option("inferSchema", "true").csv(ruta_principal+'olist_orders_dataset.csv')
df_6.write.format("delta").mode("overwrite").saveAsTable(Schema_bronze+"orders")

#Products
df_7 = spark.read.option("header", "true").option("inferSchema", "true").csv(ruta_principal+'olist_products_dataset.csv')
df_7.write.format("delta").mode("overwrite").saveAsTable(Schema_bronze+"products")

#Sellers
df_8 = spark.read.option("header", "true").option("inferSchema", "true").csv(ruta_principal+'olist_sellers_dataset.csv')
df_8.write.format("delta").mode("overwrite").saveAsTable(Schema_bronze+"sellers")

#product_category_translation
df_9 = spark.read.option("header", "true").option("inferSchema", "true").csv(ruta_principal+'product_category_name_translation.csv')
df_9.write.format("delta").mode("overwrite").saveAsTable(Schema_bronze+"product_category_translation")

```
